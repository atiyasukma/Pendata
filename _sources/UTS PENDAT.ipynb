### Impute

### Continuize
Mengubah data yang masih berbentuk kategorikal menjadi numerik agar dapat diproses oleh algoritma machine learning.


image.png


### Preprocess
Melakukan tahap praproses data, seperti normalisasi atau penyesuaian skala, sehingga data menjadi lebih siap untuk digunakan dalam pemodelan.


image.png


### Test and Score
Pada tahap Test and Score, model k-Nearest Neighbors (kNN) dievaluasi menggunakan metode cross validation dengan 10 fold. Metode ini membagi data menjadi beberapa bagian untuk diuji secara bergantian agar hasil evaluasi lebih akurat.
Berdasarkan hasil yang diperoleh, semua nilai evaluasi seperti AUC, Accuracy (CA), F1 Score, Precision, Recall, dan MCC bernilai 1.000. Hal ini menunjukkan bahwa model memiliki performa yang sangat baik karena mampu mengklasifikasikan data dengan benar.

image.png


### KNN
Pada model k-Nearest Neighbors (kNN), digunakan jumlah tetangga (k) = 5 dengan jarak Euclidean. Artinya, setiap data akan diklasifikasikan berdasarkan 5 data terdekatnya.
Pengaturan weight = uniform menunjukkan bahwa semua tetangga memiliki pengaruh yang sama dalam menentukan hasil klasifikasi.

image.png


### Confusion Matrix
Berdasarkan hasil Confusion Matrix, seluruh data berhasil diklasifikasikan dengan benar.
Data dengan kelas Subur diprediksi sebagai Subur sebesar 100%, dan data Tidak Subur juga diprediksi dengan benar sebesar 100%.

Hal ini menunjukkan bahwa model memiliki tingkat akurasi yang sangat tinggi, karena tidak terdapat kesalahan klasifikasi pada data uji.

# Modelling
Dalam proses data mining, modelling adalah tahap membangun model dari data yang sudah disiapkan. Pada tahap ini, data digunakan untuk menerapkan metode atau algoritma tertentu agar dapat menemukan pola, membuat prediksi, atau menghasilkan keputusan.

Sederhananya, modelling adalah proses mengolah data dengan teknik data mining untuk mendapatkan hasil analisis sesuai tujuan.

Beberapa kegiatan pada tahap modelling antara lain:

1. **Memilih metode atau algoritma**

   Menentukan teknik yang sesuai, misalnya klasifikasi, clustering, regresi, atau decision tree, tergantung tujuan analisis.

2. **Membagi data**

   Data biasanya dibagi menjadi data latih (training) dan data uji (testing) agar model dapat dibuat dan diuji.

3. **Menerapkan model**

   Menjalankan algoritma pada data yang sudah dipersiapkan untuk menghasilkan pola atau prediksi.

4. **Menyesuaikan parameter**

   Mengatur konfigurasi model agar hasil yang diperoleh lebih optimal.

Contohnya, jika menggunakan dataset kesehatan untuk memprediksi penyakit, tahap modelling bisa menggunakan algoritma seperti decision tree untuk menentukan apakah seseorang berisiko sakit berdasarkan data usia, tekanan darah, dan lainnya.

Jadi, modelling bertujuan membangun model dari data agar dapat menghasilkan informasi, prediksi, atau keputusan berdasarkan pola yang ditemukan dalam dataset.

## penambangan data
Data Mining atau penambangan data adalah serangkaian proses komputasi untuk menemukan pola, korelasi, dan anomali dalam kumpulan data yang besar. Melalui teknik statistik dan algoritma pembelajaran mesin (Machine Learning), proses ini bertujuan untuk mengubah data mentah menjadi informasi berharga yang dapat mendukung pengambilan keputusan strategis.

Struktur Proses Penambangan Data (CRISP-DM).
Berikut adalah penjelasan mendalam mengenai lima tahap utama dalam siklus hidup proyek penambangan data berdasarkan struktur yang Anda berikan:

### 1. Business Understanding (Pemahaman Bisnis)
Tahap awal ini berfokus pada pemahaman tujuan dan kebutuhan proyek dari sudut pandang organisasi atau pengguna.

- Menentukan Tujuan Bisnis: Langkah ini melibatkan identifikasi masalah utama yang ingin diselesaikan melalui analisis data agar hasil proyek memiliki dampak yang nyata.

- Menilai Situasi: Proses ini mencakup evaluasi ketersediaan sumber daya, perangkat lunak yang diperlukan, serta potensi hambatan yang mungkin muncul selama proyek berjalan.

- Menentukan Tujuan Data Mining: Tujuan bisnis yang bersifat umum diterjemahkan ke dalam parameter teknis data, seperti menetapkan jenis algoritma klasifikasi yang akan digunakan.

### 2. Data Understanding (Pemahaman Data)
Fokus pada tahap ini adalah pengenalan mendalam terhadap karakteristik dan kualitas data mentah yang telah dikumpulkan.

- Mengumpulkan Data: Data diekstraksi dari berbagai sumber, seperti file CSV atau basis data, untuk dijadikan bahan baku utama dalam analisis selanjutnya.

- Mendeskripsikan Data: Peneliti memeriksa jumlah instansi (baris), fitur (kolom), serta tipe data untuk memastikan formatnya sesuai dengan kebutuhan sistem.

- Eksplorasi Data: Dilakukan pengecekan awal melalui teknik visualisasi untuk menemukan tren sederhana atau hubungan antar variabel dalam dataset.

#### I. Mengumpulkan Data
Proses pengambilan dan pemuatan dataset ke dalam lingkungan analisis.

Dataset yang digunakan adalah file IRIS.csv yang terdiri dari 150 observasi. Data ini mencakup pengukuran morfologi dari tiga spesies bunga Iris berbeda. Setiap sampel memiliki 4 fitur numerik (panjang dan lebar kelopak/mahkota) dan 1 label kategori.

#### II. Mendeskripsikan Data
Identifikasi struktur, tipe data, dan definisi setiap variabel dalam dataset.

Dataset ini memiliki 5 kolom utama. Fitur sepal_length, sepal_width, petal_length, dan petal_width bertipe float (angka desimal dalam cm). Sedangkan kolom species bertipe string yang berisi nama kelas target: Iris-setosa, Iris-versicolor, dan Iris-virginica.

#### III. Eksplorasi Data
Peninjauan pola awal, distribusi, dan hubungan antar variabel.

Hasil observasi menunjukkan distribusi kelas yang seimbang (50 sampel per spesies). Secara visual, spesies Iris-setosa memiliki ukuran petal yang jauh lebih kecil dan terpisah secara linier dari spesies lain, menjadikannya fitur pembeda yang sangat kuat dalam analisis klasifikasi.

#### IV. Verifikasi Kualitas Data
Pemeriksaan kebersihan data untuk memastikan kelayakan analisis.

Verifikasi menunjukkan bahwa dataset ini memiliki kualitas yang sangat baik. Tidak ditemukan nilai yang hilang (missing values) atau data kosong (null). Penamaan label kategori juga sudah konsisten, sehingga data siap digunakan untuk pemodelan tanpa proses pembersihan (data cleaning) tambahan yang rumit.

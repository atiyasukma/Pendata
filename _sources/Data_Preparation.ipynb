# Data Preparation
Data preparation adalah tahap menyiapkan data agar siap digunakan dalam proses analisis atau pembuatan model. Setelah data dipahami pada tahap data understanding, data biasanya masih mentah sehingga perlu diproses terlebih dahulu.

Tahap ini dilakukan untuk memastikan data lebih rapi, sesuai kebutuhan, dan menghasilkan analisis yang lebih baik.

Beberapa kegiatan dalam data preparation antara lain:

1. **Memilih data yang digunakan**

     Menentukan atribut atau kolom yang relevan dengan tujuan analisis, sehingga data yang dipakai hanya yang dibutuhkan.

2. **Membersihkan data**

     Mengatasi data kosong, data ganda, kesalahan penulisan, atau nilai yang tidak sesuai agar dataset lebih akurat.

3. **Mengubah format data**

     Menyesuaikan tipe data, misalnya mengubah teks menjadi angka atau menyesuaikan format tanggal.

4. **Menggabungkan atau menyusun ulang data**

     Jika data berasal dari beberapa sumber, data dapat digabungkan menjadi satu dataset yang lengkap.

5. **Transformasi data**

     Melakukan perubahan seperti normalisasi, pengelompokan, atau pengkodean agar data lebih sesuai untuk diproses oleh sistem.

Contohnya, jika menggunakan dataset mahasiswa, pada tahap ini kita bisa menghapus kolom yang tidak diperlukan, mengisi data yang kosong, atau mengubah data kategori seperti “Laki-laki” dan “Perempuan” menjadi bentuk angka.

Jadi, data preparation bertujuan menyiapkan data yang sudah dipahami sebelumnya menjadi data yang bersih, terstruktur, dan siap dianalisis atau diproses lebih lanjut.

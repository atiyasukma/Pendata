# Business Understanding

Fokus pada tahap ini adalah mengenali data yang sudah dikumpulkan agar diketahui kondisi awalnya sebelum masuk ke proses pengolahan lebih lanjut. Pada tahap ini belum dilakukan pembersihan atau pemodelan, tetapi lebih pada memahami isi dan karakteristik data.

Data yang digunakan biasanya berasal dari berbagai sumber seperti file CSV, database, atau sumber data lain yang sesuai dengan kebutuhan analisis.

Pada tahap ini dilakukan beberapa hal, yaitu:

- Mengumpulkan data dari berbagai sumber sebagai bahan utama analisis
- Mendeskripsikan data dengan melihat jumlah baris, kolom, serta tipe data pada setiap atribut
- Melakukan eksplorasi data awal untuk melihat gambaran umum dataset

Eksplorasi tersebut biasanya dilakukan menggunakan tabel atau visualisasi sederhana untuk menemukan pola awal, hubungan antar variabel, atau kemungkinan adanya data yang tidak sesuai.

Hasil dari tahap ini menjadi dasar untuk proses selanjutnya dalam data mining.

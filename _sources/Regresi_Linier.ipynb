# Analisa Data Menggunakan Regresi Linier

## 1. Pengertian Regresi Linier
Regresi linier adalah metode analisis data yang digunakan untuk mengetahui hubungan antara variabel bebas dan variabel terikat.

Variabel bebas biasanya disebut X, sedangkan variabel terikat disebut Y.

Pada proyek ini, data yang digunakan berasal dari titik-titik di GeoGebra:
| Titik |  X |  Y |
| ----- | -: | -: |
| A     |  2 |  2 |
| B     |  4 |  3 |
| C     |  3 |  5 |
| D     |  3 |  4 |
| E     |  3 |  3 |
| F     |  4 |  5 |
| G     |  5 |  6 |

Dalam data tersebut:

     - X = variabel bebas
     - Y = variabel terikat
Tujuan regresi linier adalah mencari garis lurus terbaik yang dapat mewakili pola hubungan antara X dan Y.

## 2. Bentuk Persamaan Regresi Linier
Persamaan regresi linier sederhana adalah:


$$Y = \beta_0 + \beta_1 X$$


*Keterangan*

    - Y  = nilai yang diprediksi
    - X  = variabel bebas
    - β0 = intercept atau konstanta
    - β1 = koefisien regresi atau slope

Intercept $β_0$ adalah nilai Y ketika X bernilai 0.

Koefisien $β_1$ menunjukkan seberapa besar perubahan Y jika X naik satu satuan.

## 3. Tujuan Regresi Linier

Tujuan utama regresi linier adalah mencari garis terbaik yang memiliki kesalahan prediksi paling kecil.

Misalnya terdapat data asli:


$$(X_1​,Y_1​),(X_2​,Y_2​),...,(X_n​,Y_n​)$$


Regresi linier akan mencari garis yang paling mendekati semua titik data tersebut.

Garis regresi tidak harus melewati semua titik, tetapi garis tersebut dibuat agar jarak kesalahan antara nilai asli dan nilai prediksi menjadi sekecil mungkin.


## 4. Nilai Prediksi dan Error
Jika sudah diperoleh persamaan regresi:


$$Y=β_0​+β_1​X$$


maka setiap nilai X dapat digunakan untuk memprediksi nilai Y.

Nilai prediksi biasanya ditulis sebagai:


$$Y^$$


Error adalah selisih antara nilai Y asli dan nilai Y prediksi.


$$e=Y−Y^$$


*Keterangan:*

    - $e$  = error
    - $Y$  = nilai asli
    - $Ŷ$  = nilai prediksi

Semakin kecil error, semakin baik model regresinya.

## 5. Metode Least Squares

Regresi linier menggunakan metode least squares atau kuadrat terkecil.

Metode ini mencari garis regresi yang membuat jumlah kuadrat error menjadi sekecil mungkin.

*Rumus error kuadrat:*


$$
\sum (Y_i - \hat{Y_i})^2
$$


Keterangan:

- $\sum$ = simbol penjumlahan
- $Y_i$ = nilai data asli ke-i
- $\hat{Y_i}$ = nilai prediksi ke-i dari model regresi
- $Y_i - \hat{Y_i}$ = error atau selisih antara nilai asli dan nilai prediksi
- $(Y_i - \hat{Y_i})^2$ = error yang dikuadratkan


Artinya, setiap selisih antara nilai asli dan nilai prediksi dikuadratkan, lalu dijumlahkan.

Garis terbaik adalah garis yang memiliki total kuadrat error paling kecil.


## 6. Rumus Analitik Regresi Linier

Untuk mencari koefisien regresi secara analitik, digunakan rumus:

$$
\hat{\beta} = (X^T X)^{-1} X^T Y
$$

Keterangan:

| Simbol | Keterangan |
|---|---|
| $\hat{\beta}$ | Matriks koefisien regresi |
| $X$ | Matriks variabel bebas |
| $X^T$ | Transpose dari matriks $X$ |
| $(X^T X)^{-1}$ | Invers dari matriks $X^T X$ |
| $Y$ | Matriks variabel terikat |

Karena regresi linier memiliki intercept, maka matriks $X$ perlu ditambahkan kolom angka 1.

Contohnya, jika data $X$ adalah:

$$
2, 4, 3, 3, 3, 4, 5
$$

Maka matriks $X$ menjadi:

$$
X =
\begin{bmatrix}
1 & 2 \\
1 & 4 \\
1 & 3 \\
1 & 3 \\
1 & 3 \\
1 & 4 \\
1 & 5
\end{bmatrix}
$$

Kolom pertama berisi angka 1 untuk menghitung intercept, sedangkan kolom kedua berisi nilai variabel bebas $X$.

## 7. Regresi Linier pada GeoGebra

GeoGebra digunakan untuk memvisualisasikan titik-titik data dan membuat garis regresi.

Data titik yang digunakan adalah:

$$
L(2,2), B(4,3), C(3,5), D(3,4), E(3,3), F(4,5), G(5,6)

Di GeoGebra, titik-titik tersebut dibuat menjadi scatter plot. Setelah itu digunakan perintah:

```text
FitLine(L)
```

Perintah tersebut menghasilkan garis regresi linier terbaik berdasarkan titik-titik data.

Dari data tersebut, persamaan garis regresinya adalah:

$$
y = 1.05x + 0.4
$$

$$
Y = 0.4 + 1.05X
$$

Keterangan:

| Komponen | Nilai | Keterangan |
|---|---:|---|
| Intercept | $0.4$ | Nilai prediksi $Y$ ketika $X = 0$ |
| Koefisien regresi | $1.05$ | Besarnya perubahan $Y$ ketika $X$ naik 1 satuan |

## 8. Interpretasi Hasil Regresi

Dari persamaan:

$$
Y = 0.4 + 1.05X
$$

Dapat dijelaskan bahwa:

### Intercept

$$
\beta_0 = 0.4
$$

Artinya, ketika $X = 0$, maka nilai prediksi $Y$ adalah sebesar **0.4**.

### Koefisien Regresi

$$
\beta_1 = 1.05
$$

Artinya, setiap nilai $X$ naik sebesar 1 satuan, maka nilai $Y$ diprediksi naik sebesar **1.05 satuan**.

Karena nilai koefisien regresinya positif, maka hubungan antara $X$ dan $Y$ bersifat positif.

Dengan kata lain, semakin besar nilai $X$, maka nilai $Y$ cenderung meningkat.

## 9. Implementasi Program Python Menggunakan sklearn

Pada tahap ini, regresi linier dihitung menggunakan bahasa pemrograman Python dengan bantuan library `sklearn`. Library yang digunakan adalah `LinearRegression` dari `sklearn.linear_model`.

Data yang digunakan sama dengan data pada GeoGebra, yaitu titik A sampai G.

````python
# Import library
import pandas as pd
from sklearn.linear_model import LinearRegression

# Membuat dataset dari titik GeoGebra
df = pd.DataFrame({
    "Titik": ["A", "B", "C", "D", "E", "F", "G"],
    "X": [2, 4, 3, 3, 3, 4, 5],
    "Y": [2, 3, 5, 4, 3, 5, 6]
})

display(df)

# Menentukan variabel bebas dan variabel terikat
X = df[["X"]]
Y = df["Y"]

# Membuat dan melatih model regresi linier
model = LinearRegression()
model.fit(X, Y)

# Mengambil hasil intercept dan koefisien
intercept = model.intercept_
koefisien = model.coef_[0]

# Menampilkan hasil regresi
print(f"Intercept β0: {intercept:.2f}")
print(f"Koefisien β1: {koefisien:.2f}")
print(f"Persamaan regresi: Y = {intercept:.2f} + {koefisien:.2f}X")
````

$$
\beta_0 = 0.4
$$

$$
\beta_1 = 1.05
$$


Maka persamaan garis regresi terbaik adalah:


$$
\hat{Y} = 0.4 + 1.05X

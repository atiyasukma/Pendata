# Deployment
Tahap deployment adalah proses menerapkan hasil analisis atau model yang sudah dibuat ke penggunaan nyata. Pada tahap ini, hasil dari data mining mulai dimanfaatkan agar bisa membantu pengguna atau menyelesaikan suatu masalah.

Hal yang dilakukan pada deployment, antara lain:

1. **Menerapkan model ke data baru**

   Model digunakan untuk memproses data yang belum pernah diuji sebelumnya.

2. **Menyajikan hasil**

   Informasi dari model dapat ditampilkan dalam bentuk tabel, grafik, atau laporan.

3. **Membantu pengambilan keputusan**

   Hasil analisis dipakai sebagai bahan pertimbangan dalam menentukan tindakan.

4. **Evaluasi penggunaan**

   Model dipantau apakah masih sesuai dan bekerja dengan baik saat digunakan.

Melalui tahap ini, seluruh proses data mining yang telah dilakukan sebelumnya dapat memberikan manfaat secara langsung. Jadi, deployment bukan hanya akhir proses, tetapi juga tahap penerapan hasil analisis agar benar-benar berguna.
